**Creating our first class customer**

```python
class CustomerGikosh:
    """A class that will be a blueprint for all the customers in Gikosh
    it will contain the following states and behaviours
    - State: name, rewards_number, email, phone_number
    - Behaviours: make_purchase, return_purchase, order_online, cancle_order"""

    def __init__(self, name, rewards_number, email, phone_number):
        self.name = name
        self.rewards_number = rewards_number
        self.email = email
        self.phone_number = phone_number

    def make_purchase(self):
        print(f"Thank you {self.name} for your purchase. You have earned 200 points under reward number {self.rewards_number}")

    def return_purchase(self):
        print("Return of purchase for order 092 is approved. You will receive refund in 3-5 business days.")

    def order_online(self):
        print(f"Thank you for your online order {self.name}")

    def cancle_order(self):
        print(f"Your order for {self.name} has been cancelled.")
        
```

```python
custOne = CustomerGikosh("Hamza", 1010, "hamza@gmail.com", "254748896675")
```

```python
print(custOne.name)
print(custOne.rewards_number)
print(custOne.email)
print(custOne.phone_number)
```

```python
custOne.make_purchase()
```

**Inheritance and Polymorphism
- inheritance allows you to create a new class(child/sub-class) that inherits from an existing class( parent/base class)
- polymorphism allows object of different classes to be treated as object of the same parent class.

```python
class Animal:
    def __init__(self, name, color, breed):
        self.name = name
        self.color = color
        self.breed = breed

    def speak(self):
        return f"{self.name} makes a sound"



class Dog(Animal):

    def speak(self):
        return f"{self.name} says Woof!"

    def run(self):
        return f"{self.name} is really fast!"

dog = Dog("Jimmy", "Brown", "G.shepherd")

print(dog.name)
print(dog.color)
print(dog.breed)
print(dog.run())
print(dog.speak())

class Cat(Animal):

    def speak(self):
        return super().speak()

cat = Cat("Rubsky", "Black", "Bengal")

print(cat.speak())
        
        
```

```python
class BankAccount:
    """This is a class used to instantiate Bank Customers
    State: balance"""

    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        self.balance -= amount 

# An empty class inheriting from the bankaccount class

class SavingAccount(BankAccount):
    # Constructor for savings account with additional arguments

    def __init__(self, balance, interest_rate):
        #call the Parent Constructor using ClassName.__unit__()
        #Self is a savings account but also a bankaccount
        BankAccount.__init__(self,balance)

        #super().__init__(balance)

        #Add more funtionality
        self.interest_rate = interest_rate 

    def compute_interest(self, n_periods = 2):
         return (self.balance * ((1 + self.interest_rate) ** n_periods)) - self.balance # this is for compound interest
        #return self.balance * self.interest_rate * n_periods # this is for simple interest
```

```python
acc = SavingAccount(4000, 0.03)
acc.interest_rate
acc.compute_interest()
```
